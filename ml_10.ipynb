```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[0], cmap='gray')
plt.title("Normalized Image")
plt.axis('off')
plt.show()
```

```python
original_image = x_train[0].copy()
x_train = x_train / 255.0
x_test = x_test / 255.0
print("Before Normalization:")
print(original_image[:5, :5])
print("After Normalization:")
print(x_train[0][:5, :5])
```

```python
import pandas as pd
df = pd.DataFrame(x_train[0])
print(df)
```

```python
model = tf.keras.Sequential([ tf.keras.layers.Flatten(input_shape=(28, 28)), tf.keras.layers.Dense(128, activation='relu'), tf.keras.layers.Dense(10, activation='softmax') ])
```

```python
model.summary()
```

```python
model.compile( optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'] )
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
# Generate predictions on the test dataset
prediction = model.predict(x_test)

# Find the class with the highest probability for the first test sample
predicted_label = np.argmax(prediction[0])
print("Predicted Class:", class_names[predicted_label])
```
