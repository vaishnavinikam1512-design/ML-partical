```python
employee_num = employee.select_dtypes(include=np.number)
employee_num.head(3)
```

```python
employee_num = employee.select_dtypes(include=[np.number])
employee_num.head(3)
```

```python
from sklearn.preprocessing import LabelEncoder

# Select categorical columns from the 'employee' DataFrame
employee_dummies = employee.select_dtypes(include='object').copy()

# Apply LabelEncoder to each categorical column
for column in employee_dummies.columns:
    employee_dummies[column] = LabelEncoder().fit_transform(employee_dummies[column])

employee_dummies.head(3)
```

```python
employee_combined= pd.concat([employee_num,employee_dummies],axis=1)
employee_combined.head(3)
```

```python
from sklearn.model_selection import train_test_split

x = employee_combined.drop(["Churn"], axis = 1)
y = employee_combined["Churn"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 231)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(x_train,y_train)
```

```python
train = pd.concat([x_train,y_train],axis=1)
train.head()
```

```python
independent_varibale = list(train.columns)
independent_varibale
```

```python
train["Predicted"] = dt.predict_proba(x_train)[:, 1]
train.head()
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
from sklearn.model_selection import GridSearchCV

# Parameter grid
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10]
}

churn = ['No', 'Yes']

fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(5,4), dpi=300)

tree.plot_tree(
    dt,
    feature_names=independent_varibale,
    class_names=churn,
    filled=True,
    node_ids=True,
    fontsize=2
)

# Grid Search
grid_search = GridSearchCV(dt, param_grid, cv=5)

grid_search.fit(x, y)

print(grid_search.best_params_)
print(grid_search.best_score_)
```
