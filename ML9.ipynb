```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('/content/Country-data.csv')
df
```

```python
print("First 5 records:")
print(df.head())
```

```python
x = df.drop('country', axis=1)
x
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
print(x_scaled[:5])
```

```python
x_scaled = pd.DataFrame(
scaler.fit_transform(x),
columns=x.columns
)
print(x_scaled.head())
```

```python
kmeans = KMeans(n_clusters = 3, random_state=42, n_init=10)
```

```python
kmeans.fit(x_scaled)
```

```python
df['cluster'] = kmeans.labels_
print("\nCountries with Cluster:")
print(df[['country','cluster']].head(15))
```

```python
print("\nNumber of Countries in each Cluster:")
print(df['cluster'].value_counts().sort_index())
```

```python
new_sample = pd.DataFrame(
[[10,40,7,35,30000,5,75,2,15000]],
columns=x.columns
)
new_sample_scaled = pd.DataFrame(
scaler.transform(new_sample),
columns=x.columns
)
new_cluster = kmeans.predict(new_sample_scaled)
```

```python
print("\nNew Country Sample:")
print(new_sample)
print("\nnew country belongs to cluster:")
new_cluster[0]
```
